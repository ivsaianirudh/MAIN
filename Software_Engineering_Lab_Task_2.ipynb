```python
# Read Single Input from File
try:
    with open("single_input.txt", "r") as f:
        num = int(f.read())
        print("Number from file:", num)
except FileNotFoundError:
    print("File not found")
```

```python
with open("multiple_inputs.txt", "w") as f:
    f.write("10\n20\n30\n40")
```

```python
# Read Multiple Inputs from File
try:
    with open("multiple_inputs.txt", "r") as f:
        numbers = f.readlines()
        total = 0
        for n in numbers:
            total += int(n.strip())
        print("Sum of numbers:", total)
except FileNotFoundError:
    print("File not found")
```

```python
# Debugging Example
try:
    num = int(input("Enter a number: "))
    print("Square:", num * num)
except ValueError:
    print("Invalid input! Please enter a number.")
```
